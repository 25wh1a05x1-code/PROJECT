```python
from datetime import datetime, date, timedelta
now = datetime.now()
print("Now:", now)
today = date.today()
print("Today's Date:", today)
dob = date(2000, 1, 1)
print("DOB:", dob)
yesterday = today - timedelta(days=1)
tomorrow = today + timedelta(days=1)
print("Yesterday:", yesterday)
print("Tomorrow:", tomorrow)
formatted = now.strftime("%d-%m-%Y %H:%M:%S")
print("Formatted DateTime:", formatted)
```

```python
import random
print("Random Int:", random.randint(1, 100))
print("Random Float:", random.random())
colors = ['red', 'green', 'blue']
print("Random Color:", random.choice(colors))
random.shuffle(colors)
print("Shuffled List:", colors)
print("Sample:", random.sample(range(1, 50), 5))
```

```python
from collections import defaultdict
marks = defaultdict(lambda: 0)
marks['Math'] = 90
marks['Science'] = 80
print("English Marks (not set):", marks['English'])
```

```python
from collections import namedtuple
Point = namedtuple('Point', 'x y')
p1 = Point(10, 20)
print("Point:", p1)
print("x:", p1.x, "y:", p1.y)
```

```python
from collections import deque
dq = deque([1, 2, 3])
dq.append(4)
dq.appendleft(0)
dq.pop()
dq.popleft()
print("Deque after operations:", dq)
```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)
print_names("Alice", "bob", "Charlie")
```

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")
print_info(name="Alice", age=25, city="Delhi")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)
show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
def multiply(a, b):
    return a * b
def square(n):
    return multiply(n, n)
print("Square of 4:", square(4))
```

```python
def f(x):
    if x >= 0:
        return x ** 2
    else:
        return -x
for val in [-3, -1, 0, 2, 5]:
    print(f'f({val:3d}) = {f(val)}')
```

```python
score = 73
if score >= 90:
    grade = 'A (Excellent)'
elif score >= 80:
    grade = 'B (Good)'
elif score >= 70:
    grade = 'C (Average)'
elif score >= 60:
    grade = 'D (Below Average)'
else:
    grade = 'F (Fail)'
print(f'Score: {score} → Grade: {grade}')
```

```python
x = 15
if x > 10 and x < 20:
    print(f'{x} is between 10 and 20')
if x < 5 or x > 12:
    print(f'{x} is outside the range 5–12')
if not x == 10:
    print(f'{x} is not equal to 10')
```

```python
n = 36
if n % 2 == 0:
    print(f'{n} is Even')
else:
    print(f'{n} is Odd')
if n % 3 == 0 and n % 5 == 0:
    print(f'{n} is divisible by both 3 and 5')
elif n % 3 == 0:
    print(f'{n} is divisible by 3 only')
elif n % 5 == 0:
    print(f'{n} is divisible by 5 only')
```

```python
n = 10
total = 0
for i in range(1, n + 1):
    total += i
formula_result = n * (n + 1) // 2
print(f'Sum (loop)    : {total}')
print(f'Sum (formula) : {formula_result}')
```

```python
n = 6
factorial = 1
for i in range(1, n + 1):
    factorial *= i
    print(f'  Step {i}: {i}! so far = {factorial}')
print(f'\n6! = {factorial}')
```

```python
a, b = 0, 1
limit = 100
print('Fibonacci numbers below', limit, ':')
while a < limit:
    print(a, end='  ')
    a, b = b, a + b
```

```python
print('Multiplication Table (1–5)\n')
print('   ', end='')
for col in range(1, 6):
    print(f'{col:4}', end='')
print()
print('─' * 25)
for row in range(1, 6):
    print(f'{row:2} |', end='')
    for col in range(1, 6):
        print(f'{row*col:4}', end='')
    print()
```
